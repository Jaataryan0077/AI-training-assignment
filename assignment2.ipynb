```python
# TASK 1
from transformers import AutoTokenizer, AutoModelForCausalLM

model = "meta-llama/Llama-3.2-1B-Instruct"
tokenizer = AutoTokenizer.from_pretrained(model)
model = AutoModelForCausalLM.from_pretrained(model)

print("Tokenizer:\n", tokenizer)
print("Model:\n", model)
```

```python
from transformers import pipeline

chatPipe = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer
)
```

```python
while True:
    user_input = input("Ask Anything: ")
    print("HUMAN:", user_input)

    if user_input.lower() in ["quit", "bye", "exit"]:
        print("AI ASSISTANT:Bye,Stay healthy.")
        break

    prompt = f"""You are an AI Healthcare Assistant that helps doctors summarize patient reports, answer clinical questions, and generate discharge summaries.

Your job is to answer medical questions in a helpful manner.

Rules:
1)Answer the user's question directly.
 2)Be clear and professional.
 3)If you are unsure, say "I don't know" and recommend consulting a healthcare professional. Do not ask the user to repeat their question.


User Question:
{user_input}

Assistant:
"""

    response = chatPipe(
        prompt,
        max_new_tokens=100,
        do_sample=True,
        temperature=0.7
    )[0]["generated_text"]

    print("AI Response:\n", response[len(prompt):].strip())
```
